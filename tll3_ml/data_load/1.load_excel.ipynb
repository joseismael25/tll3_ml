```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.read_excel('data/dataset_ventas.xlsx')
df.columns = df.columns.str.strip()

print("Columnas que hay:", df.columns.tolist())

le = LabelEncoder()

# Codificamos categoria, que SI existe
df['categoria_codificada'] = le.fit_transform(df['categoria'])

# Esto te muestra qué número le puso a cada categoría
print(dict(zip(le.classes_, le.transform(le.classes_))))

print(df[['categoria', 'categoria_codificada']].head(10))
print(df.head())
```

```python
from sklearn.preprocessing import LabelEncoder

# Identificar columnas de texto
columnas = df.select_dtypes(include=['object']).columns

# Aplicar LabelEncoder
for columna in columnas:
    le = LabelEncoder()
    df[columna] = le.fit_transform(df[columna].astype(str))

# Aplicar One-Hot Encoding
df = pandita.get_dummies(df, dtype=int)

display(df.head())



from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Seleccionar solamente las columnas numéricas
columnas_numericas = df.select_dtypes(include=['number']).columns

# Normalización
normalizador = MinMaxScaler()

df_normalizado = df.copy()

df_normalizado[columnas_numericas] = normalizador.fit_transform(
    df[columnas_numericas]
)

# Estandarización
estandarizador = StandardScaler()

df_estandarizado = df.copy()

df_estandarizado[columnas_numericas] = estandarizador.fit_transform(
    df[columnas_numericas]
)

# Mostrar resultados
print("Datos normalizados:")
display(df_normalizado.head())

print("Datos estandarizados:")
display(df_estandarizado.head())




from imblearn.over_sampling import SMOTE

# Eliminar la columna fecha
df_sin_fecha = df.drop('fecha', axis=1)

# Separar variables predictoras y objetivo
X = df_sin_fecha.drop('categoria_codificada', axis=1)
y = df_sin_fecha['categoria_codificada']

# Aplicar SMOTE
smote = SMOTE(random_state=42)

X_balanceado, y_balanceado = smote.fit_resample(X, y)

# Mostrar resultados
print("Antes de SMOTE:")
print(y.value_counts())

print("\nDespués de SMOTE:")
print(y_balanceado.value_counts())

print("\nDimensiones de los datos balanceados:")
print(X_balanceado.shape)


```
