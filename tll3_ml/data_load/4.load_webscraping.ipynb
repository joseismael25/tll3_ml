```python

import requests
import pandas as pandita
from bs4 import BeautifulSoup

# URL de la página web
url = 'https://www.scrapethissite.com/pages/simple/'

# Consumir la página
respuesta = requests.get(url)

# Verificar conexión
print("Código de respuesta:", respuesta.status_code)

# Analizar el HTML
soup = BeautifulSoup(respuesta.text, 'html.parser')

# Extraer los países
paises = soup.find_all('div', class_='country')

lista_paises = []

for pais in paises:
    nombre = pais.find('h3', class_='country-name').text.strip()
    capital = pais.find('span', class_='country-capital').text.strip()
    poblacion = pais.find('span', class_='country-population').text.strip()
    area = pais.find('span', class_='country-area').text.strip()

    lista_paises.append({
        'pais': nombre,
        'capital': capital,
        'poblacion': poblacion,
        'area': area
    })

# Crear DataFrame
df = pandita.DataFrame(lista_paises)

# Mostrar resultados
display(df.head())

print("Total de países:", len(df))
```

```python

# Información del DataFrame
df.info()

# Estadísticas
display(df.describe())

# Mostrar los 10 países más poblados
df['poblacion'] = pandita.to_numeric(df['poblacion'])

display(
    df.sort_values(
        by='poblacion',
        ascending=False
    ).head(10)
)
```
